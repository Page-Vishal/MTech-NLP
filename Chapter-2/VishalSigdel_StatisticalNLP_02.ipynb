# Predict the Next Word Using Conditional Probability

Build bigram counts from a training corpus, compute conditional probabilities
$P(w_2 \mid w_1) = \dfrac{Count(w_1, w_2)}{Count(w_1)}$, then ask the user for a
word and display all possible next words ranked by probability along with the
top prediction.

**Author:** Vishal Sigdel

## Step 1: Training Corpus

The corpus used to learn bigram statistics.

```python
training_corpus = [
    "I like NLP",
    "I like Python",
    "I like coffee",
    "I love NLP",
    "You like coffee",
]
```

## Step 2: Build Bigram Counts

Count($w_1, w_2$) for every adjacent word pair in the corpus.

```python
bigram_counts = {}
for sentence in training_corpus:
    tokens = sentence.split()
    for w1, w2 in zip(tokens, tokens[1:]):
        bigram_counts.setdefault(w1, {})
        bigram_counts[w1][w2] = bigram_counts[w1].get(w2, 0) + 1

for w1, next_words in bigram_counts.items():
    for w2, c in next_words.items():
        print(f"{w1} -> {w2} : {c}")
```

## Step 3: Conditional Probabilities

$P(w_2 \mid w_1) = \dfrac{Count(w_1, w_2)}{Count(w_1)}$, computed for every bigram.

```python
def conditional_probabilities(word):
    candidates = bigram_counts.get(word, {})
    total = sum(candidates.values())
    if total == 0:
        return {}
    return {w: c / total for w, c in candidates.items()}

for w1 in bigram_counts:
    for w2, p in conditional_probabilities(w1).items():
        print(f"P({w2} | {w1}) = {p:.2f}")
```

## Step 4: Ask User for a Word

```python
print("Enter a word: ")
query_word = input().strip()
print(query_word)
```

## Step 5: Display Possible Next Words and Probabilities

```python
probs = conditional_probabilities(query_word)
ranked = sorted(probs.items(), key=lambda kv: kv[1], reverse=True)

if not ranked:
    print(f"No bigrams found starting with \"{query_word}\"")
else:
    print("Possible next words:")
    for w, p in ranked:
        print(f"{w} : {p:.2f}")
```

## Step 6: Predict Word with Highest Conditional Probability

```python
if ranked:
    predicted = ranked[0][0]
    print(f"\nPredicted next word: {predicted}")
```

## Bonus: Top Three Predictions

```python
if ranked:
    print("Top Predictions")
    for i, (w, p) in enumerate(ranked[:3], 1):
        print(f"{i}. {w} ({p:.2f})")
```
